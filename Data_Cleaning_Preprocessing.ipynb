```python
text_columns = cleaned_df.select_dtypes(include="object").columns

for col in text_columns:
    cleaned_df[col] = cleaned_df[col].str.strip()
```

```python
numeric_columns = ["Postal Code", "Sales", "Quantity", "Discount", "Profit"]

for col in numeric_columns:
    cleaned_df[col] = pd.to_numeric(cleaned_df[col], errors="coerce")
```

# Data Cleaning and Preprocessing – Sample Superstore

## Project Objective

The objective of this project is to clean and preprocess the Sample Superstore dataset using Python and Pandas. The dataset was inspected for missing values, duplicate records, inconsistent text formatting, incorrect data types, and invalid numerical values.

## Tools Used

- Python
- Pandas
- NumPy
- Google Colab

## Dataset Overview

The original dataset contains 9,994 rows and 13 columns. It contains information about shipping mode, customer segment, location, product categories, sales, quantity, discount, and profit.

## Data Quality Checks

The following checks were performed:

1. Dataset structure and dimensions
2. Missing values
3. Duplicate records
4. Categorical/text consistency
5. Extra whitespace
6. Numerical data types
7. Negative numerical values
8. Final validation

## Cleaning Performed

- Duplicate rows were removed.
- Text columns were checked and stripped of leading/trailing whitespace.
- Numeric columns were converted/validated as numeric data types.
- Missing values were checked.
- Categorical fields were checked for inconsistent formatting.
- Negative profit values were retained because they represent legitimate loss-making transactions.

## Final Result

The cleaned dataset contains 9,977 rows and 13 columns.

Final validation confirmed:

- 0 duplicate rows
- 0 missing values
- 13 columns
- Correct numeric data types

The cleaned dataset was exported as `superstore_cleaned.csv`.
